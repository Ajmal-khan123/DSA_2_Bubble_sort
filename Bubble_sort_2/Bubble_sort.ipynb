### Bubble Sort__

```python
array = [3, 2, 8, 1, 5, 9]

n = len(array)
print(f'the length of the array {n}')

for i in range(n-1):  # outer loop 
    for j in range(n-i-1):  # inner loop
        if array[j] > array[j+1]: # comparison 
            array[j], array[j+1] = array[j+1], array[j] # swaping 

            
print(f'the i = {i}')
print(f'the j = {j}')
print(f'My sorted array {array}')
```

#### Bubble sort improvement

```python
array = [7, 3, 9, 12, 11]

n  = len(array)
for i in range(n-1):
    swap = False
    for j in range(n-i-1):
        if array[j] > array[j+1]:
            array[j], array[j+1] = array[j+1], array[j]
            swap = True
    if not swap:
        break
print(f'The sorted array {array}')
```
